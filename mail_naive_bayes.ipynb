```python
# ⭐ Project 5: Spam Email Detection - Naive Bayes

import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score,classification_report

data = pd.read_csv("spam.csv")
df1 = pd.DataFrame(data)
df = df1.copy()
df
```

```python
# EDA process

df["Category"] = df["Category"].map({"ham":"not spam","spam":"spam"})
print(df.head())
print(df.shape)
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```python
# Model Train Test Split

x = df["Message"].values
y = df["Category"].values
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.20,random_state=42)

model_cv = CountVectorizer()
x_train_cv = model_cv.fit_transform(x_train)
x_test_cv = model_cv.transform(x_test)

model = MultinomialNB()
model.fit(x_train_cv,y_train)


```

```python
# Model Train

y_pred = model.predict(x_test_cv)
accuracy = accuracy_score(y_test,y_pred)

print("The Model Prediction :",y_pred[0])
print(f"The model Accuracy   : {accuracy*100:.2f}")
```

```python
# New Input Testing

new_mail = input("Enter your Mail : ")
new_mail_cv = model_cv.transform([new_mail])
y_pre = model.predict(new_mail_cv)
print("\n")
print("-"*50)
print("   My model identified this email as :",y_pre[0])
print("-"*50)
```
